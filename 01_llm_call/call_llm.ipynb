```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it")

response = llm.invoke("How many moons does Jupiter have?")
print(response.text)
```

```python
response = llm.invoke([
    ["system", "You are a helpful assitstant that answers in 2 lines that rhyme"],
    ["human", "How many moons does Jupiter have?"]
])
print(response.text)
```

```python
prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)

response = llm.invoke(prompt)
print(response.text)
```

```python
prompt = "Write a single tagline for a quirky coffee shop. Only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=2)

response = llm.invoke(prompt)
print(response.text)
```

Calling LLM using Groq Model

```python
!pip install langchain_groq
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(model="qwen/qwen3-32b", reasoning_format="parsed")

messages = [
    ("system", "You are a helpful assitstant that answers in 2 lines that rhyme"),
    ("human", "How many moons does Jupiter have?"),
]

response = llm.invoke(messages)
print(response.content)
```
